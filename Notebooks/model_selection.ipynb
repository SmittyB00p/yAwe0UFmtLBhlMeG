# Model Selection Using LazyPredict

Using LazyPredict we will see what kinds of models are suggested for use with all the predictors, a subset of the 4 predictors that have over a 5% correlation with the target variable, and just one predictor.

```python
from lazypredict.Supervised import LazyClassifier
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings(action='ignore')
```

```python
df = pd.read_csv('../data/ACME-HappinessSurvey2020_shrunk.csv')

df.head()
```

```python
target_variable = df['customer satisfaction']
X = df.drop(columns='customer satisfaction', axis=1)

X.head()
```

```python
# seed = random.randint(1000, 9999)
seed = 5249
# print(seed)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    target_variable,
    test_size=.2,
    random_state=seed
)
```

```python
classifiers = LazyClassifier(verbose=0,
                             ignore_warnings=True,
                             random_state=seed,
                             custom_metric=None
)

models, preds = classifiers.fit(X_train, X_test, y_train, y_test)
models
```
